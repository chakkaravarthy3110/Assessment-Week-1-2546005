```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel("AAPL.xlsx")
```

```python
df.head()
```

```python
df.columns
```

```python
df.dtypes
```

```python
df["date"] = pd.to_datetime(df["Date"])
```

```python
df['date'].dtype
```

```python
df = df.sort_values("date")
```

```python
df["Date"] = pd.to_datetime(df["Date"])
df
```

```python
plt.figure(figsize=(16, 8))

plt.plot(df.index, df["Open"], label="Open")
plt.plot(df.index, df["High"], label="High")
plt.plot(df.index, df["Low"], label="Low")
plt.plot(df.index, df["Close"], label="Close")

plt.title("Shopify Stock OHLC Prices")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(16,8))

plt.plot(df["date"],df["Volume"])
plt.title("Apple Volume Analysis")
plt.xlabel("Date")
plt.ylabel("Volume of trade shares")
plt.show()
```

```python
df["MA20"] = df["Close"].rolling(window=20).mean()

df["MA50"] = df["Close"].rolling(window=50).mean()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df.index, df["Close"], label="Daily Close")
plt.title("Apple Stock closing price trend")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```
